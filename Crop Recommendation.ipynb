Data Preprocessing

```python
x = data.drop('label',axis=1)
y = data['label']
```

```python
x.info()
```

```python
y.info()
```

```python
from sklearn.model_selection import train_test_split,cross_val_score
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
models =  {
    'LogisticRegression': LogisticRegression(max_iter=1000),
    'DecisionTreeClassifier': DecisionTreeClassifier(),
    'RandomForestClassifier': RandomForestClassifier(),
    'KNeighborsClassifier':KNeighborsClassifier()
}

model_accuracy = {}

for name, model in models.items():
    model.fit(x_train,y_train)
    y_pred = model.predict(x_test)
    accuracy = accuracy_score(y_test,y_pred)
    print(name,'accuracy_score:',accuracy)
    model_accuracy[name] = accuracy
```

```python


real_best_model = max(model_accuracy,key = model_accuracy.get)
best_model = models[real_best_model]
```

```python
print('Best Model : ',best_model)
print("Best Accuracy:", model_accuracy[real_best_model])
```

```python
results = {}

for name, model in models.items():
    cv_scores = cross_val_score(model, x, y, cv=5)
    results[name] = cv_scores.mean()
    print(f"{name} CV Accuracy: {cv_scores.mean():.4f}")

# Best model name
best_model_name = max(results, key=results.get)
print("\nBest Model:", best_model_name)
```
